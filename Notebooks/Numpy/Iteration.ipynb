# **Iteration**

````python
import numpy as np
````

# **Using `for` loop**

### 1D Array

````python
arr=np.array([1,2,2,3,4,5])
for i in arr:
    print(i)

for i in range(len(arr)):
    print(arr[i])
````

### 2D Array

````python
arr=np.array([[1,2,3],[4,5,6]])

for i in arr:
    for j in i:
        print(j)
    print()
    
for i in range(len(arr)):
    for j in range(len(arr[i])):
        print(arr[i,j],end='')
    print('')
````

### 3D Array

````python
arr=np.arange(12).reshape(2,3,2)

# for i  in arr:
#     for j in i:
#         for k in j:
#             print(k)

for i in range(len(arr)):
    for j in range(len(arr[i])):
        for k in range(len(arr[i,j])):
            print(arr[i,j,k])
````

## **Using `nditer()` Function**

### 1D, 2D and 3D Array


````python
arr=np.array([1,2,2,3,4,5])
arr=np.array([[1,2,3],[4,5,6]])
arr=np.arange(12).reshape(2,3,2)

for i  in np.nditer(arr, flags=None, op_flags=None, order='K', casting='safe',op_dtypes=None, itershape=None, buffersize=0):
    print(i)
````

## **`ndenumerate(arr)`** : To Print Array with thier respective index.

````python
for i,v in np.ndenumerate(arr):
    print(i,v)
````

---

## 🔹 What is `nditer`?

> `np.nditer` is an efficient multi-dimensional iterator used for element-wise iteration with fine control over memory, order, and broadcasting.

---

## 🔹 Basic Syntax

```python
np.nditer(op, flags=None, op_flags=None, order='K', casting='safe',
          op_dtypes=None, itershape=None, buffersize=0)
```

---

## 📊 `nditer` Parameters Explained

### 1️⃣ `op`

| Parameter | Description                             |
| --------- | --------------------------------------- |
| `op`      | Array or list of arrays to iterate over |

Example:

```python
np.nditer(a)
np.nditer([a, b])
```

---

### 2️⃣ `flags`

Controls iterator behavior.

| Flag              | Meaning                         |
| ----------------- | ------------------------------- |
| `'c_index'`       | Track C-order index             |
| `'f_index'`       | Track Fortran-order index       |
| `'multi_index'`   | Enable multi-dim index          |
| `'external_loop'` | Yield chunks instead of scalars |
| `'buffered'`      | Enable buffering                |
| `'zerosize_ok'`   | Allow zero-size arrays          |
| `'reduce_ok'`     | Allow reduction ops             |
| `'grow_inner'`    | Grow inner loop                 |

Example:

```python
np.nditer(a, flags=['multi_index'])
```

---

### 3️⃣ `op_flags`

Controls read/write access for operands.

| Flag             | Meaning                |
| ---------------- | ---------------------- |
| `'readonly'`     | Read-only access       |
| `'readwrite'`    | Read & write           |
| `'writeonly'`    | Write-only             |
| `'allocate'`     | Allocate output array  |
| `'no_broadcast'` | Disable broadcasting   |
| `'contig'`       | Require contiguous     |
| `'aligned'`      | Require aligned memory |

Example:

```python
np.nditer(a, op_flags=['readwrite'])
```

---

### 4️⃣ `order`

Iteration order.

| Value | Meaning                     |
| ----- | --------------------------- |
| `'C'` | Row-major                   |
| `'F'` | Column-major                |
| `'A'` | Any order                   |
| `'K'` | Keep memory order (default) |

---

### 5️⃣ `casting`

Type-casting rules.

| Value         | Meaning                 |
| ------------- | ----------------------- |
| `'no'`        | No casting              |
| `'equiv'`     | Only byte-order changes |
| `'safe'`      | Safe casts only         |
| `'same_kind'` | Same kind casts         |
| `'unsafe'`    | Any cast                |

---

### 6️⃣ `op_dtypes`

| Parameter   | Description              |
| ----------- | ------------------------ |
| `op_dtypes` | Force dtype for operands |

Example:

```python
np.nditer(a, op_dtypes=['float64'])
```

---

### 7️⃣ `itershape`

| Parameter   | Description              |
| ----------- | ------------------------ |
| `itershape` | Override iteration shape |

Example:

```python
np.nditer(a, itershape=(2,3))
```

---

### 8️⃣ `buffersize`

| Parameter    | Description              |
| ------------ | ------------------------ |
| `buffersize` | Size of temporary buffer |

Used with `'buffered'` flag.

Example:

```python
np.nditer(a, flags=['buffered'], buffersize=4)
```

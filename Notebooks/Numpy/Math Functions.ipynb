# **Math Functions**

```python
import numpy as np
arr=np.array([2,4,6,8,10])
```

## **Square Root**

```python
np.sqrt(arr)
```

## **Absolute Value**

```python
np.abs(arr)
```

## **Power**

```python
np.power(arr,2)
```

## **Exponential**

```python
np.exp(arr)
```

## **Natural Log**

```python
np.log(arr)
```

## **Sine**

```python
np.sin(arr)
```

## **Cosine**

```python
np.cos(arr)
```

## **Tangent**

```python
np.tan(arr)
```

### **1D Array**

````python
arr1D=np.array([1,2,3])
arr1D.ndim
arr1D.shape
````

### **2D Array**


````python
arr2D=np.array([[1,2,3],
                [1,2,3]])
arr2D.ndim
arr2D.shape
````

### **3D Array**


````python
arr3D=np.array([[[1,2,3],
                 [1,2,3],
                 [1,2,3]],
                 [[1,2,3],
                  [1,2,3],
                  [1,2,3]]])
arr3D.ndim
arr3D.shape
````

### **nD Array**


````python
arrND=np.array([1,2,3],ndmin=6)
arrND.ndim
arrND.shape
````

> While Creating Multiple Dimensional make sure all Arrays must be same size

## **Array Creation Functions**
### **Zeros Array**
```python
np.zeros((shape), dtype=float)
```


````python
arr_zero=np.zeros((3,3,3))
arr_zero
````

### **Ones Array**
```python
np.ones((shape), dtype=float)
```


````python
arr_ones=np.ones((3,3,3))
arr_ones
````

### **Empty Array**
```python
np.empty((shape), dtype=float)
```


````python
arr_empty=np.empty((3,3,3))
arr_empty
````

### **Array Range**
```python
np.arange(start, stop, step)
```


````python
arr_range=np.arange(10)
arr_range
````

### **Identity Matrix**
```python
np.eye(N, M=None, k=0)
```
* (1s on diagonal, 0s elsewhere).
* `N` → number of rows
* `M` → number of columns (optional)
* `k` → diagonal index (0 = main diagonal)


````python
arr_diagonal=np.eye(5,5)
arr_diagonal
````

## **linspace**
##### Creates evenly spaced numbers over a fixed interval.
```python
np.linspace(start, stop, num=50)
```
* `start` → starting value
* `stop` → ending value (included)
* `num` → number of values (default 50)

````python
arr_lin=np.linspace(0,10,num=5)
arr_lin
````

# **Random Functions**

````python
import numpy as np

# All random functions live inside:
# np.random
````

## **`rand()`** 
* Generates **random floats between 0 and 1**
* Uniform Distribution
* Values are in **[0.0, 1.0]**

```python
np.random.rand(d0,d1,...dn)
```
`d0, d1,...` -> Dimensions (No Tuple Pass Directly)


````python
np.random.rand(3)
np.random.rand(3,3)
np.random.rand(3,3,3)
````

## **`randn()`**

* Generates random numbers from **standard normal distribution**
* Mean ≈ `0`, Standard Deviation ≈ `1`
* Can return **Positive or Negative (Close to Zero)** values

```python
np.random.randn(d0,d1,...dn)
```
`d0, d1,...` -> Dimensions (No Tuple Pass Directly)

````python
np.random.randn(3)
np.random.randn(3,3)
np.random.randn(3,3,3)
````

## **`ranf()`**

* Generates random floats in **[0.0, 1.0]**
* Similar to **`rand()`**, but accepts **shape as a tuple**
```python
np.random.ranf(size=None)
```
`size` → int or tuple (array shape)
#### **`np.random.random(size=())`** is also same

````python
np.random.ranf(3)
np.random.ranf((3,3))
np.random.ranf((3,3,3))
````

## **`randint()`**

* Generates **random integers** in a given range
* Lower bound **included**, upper bound **excluded**

```python
np.random.randint(low, high=None, size=None, dtype=int)
```
`low` → minimum value (inclusive)

`high` → maximum value (exclusive)

`size` → number or shape of output

`dtype` → data type (optional)

````python
np.random.randint(1,10)
np.random.randint(1,10,5)
np.random.randint(1,10,(3,3))


````

## **`choice()`**
* Randomly select values from an array
```python
np.random.random([],size=None)
```

````python
np.random.choice([1,2,3],size=2)
````

## **`shuffle()`**
* Shuffles array **in-place**
```python
np.random.shuffle(array)
```

````python
arr=np.array([1,2,3,4])
np.random.shuffle(arr)
````

## **`permutation()`**
* Returns a **shuffled copy**
```python
np.random.permutation(arr)
```

````python
np.random.permutation(arr)
````

## **`seed()`**
* Sets seed for **reproducibility**
* Fixes the random number generator
* Ensures **same output every time** (reproducibility)
```python
np.random.seed(n)
```


````python
np.random.seed(42)
np.random.rand(3)
````

## **`normal()`**
* Generates numbers from **Normal (Gaussian) distribution**
* Mean = `loc`, Standard deviation = `scale`
```python
np.random.normal(loc=0.0, scale=1.0, size=None)
```
📌 Used in ML, statistics, noise simulation.

````python
np.random.seed(42)
np.random.normal(size=5)
````

## **`uniform()`**
* Generates numbers **uniformly distributed**
* All values have equal probability
```python
np.random.uniform(low=0.0, high=1.0, size=None)
```
📌 Used when no bias is assumed.

````python
np.random.seed(42)
np.random.uniform(1, 10, 5)
````

## **`binomial()`**
* Simulates **binary outcomes** (success / failure)
* Used in **Bernoulli trials**
```python
np.random.binomial(n, p, size=None)
```
`n` → number of trials

`p` → probability of success

`size` → output shape

📌 Used in A/B testing, coin tosses, experiments.

````python
np.random.seed(42)
np.random.binomial(n=10, p=0.5, size=5)
````

## **`poisson()`**
* Model **Event counts over a fixed interval** 
* Used for **rare events**
```python
np.random.poisson(lam=1.0 size=None)
```
`lam` → average number of events

`size` → output shape

📌 Used for
* Calls per hour
* Accidents per day
* Server requests

````python
np.random.seed(42)
np.random.poisson(lam=3, size=5)
````

##### 1️⃣ Reproducible random matrix

* Set seed = `42`
* Generate a **4×4** matrix from **uniform distribution [0,1)**
* Replace:

  * values `< 0.3` → `0`
  * values `>= 0.3` → `1`


````python
np.random.seed(42)
arr=np.random.uniform(0,1,size=(4,4))
arr[arr>=0.3]=1
arr[arr<0.3]=0
arr.astype(int)

# OR
result=(arr>=0.3).astype(int)
result
````

##### 2️⃣ Normal distribution filtering

Generate **1000 samples** from a **normal distribution**:

* mean = `50`
* std = `10`
* Keep **only values between 40 and 60**
* Return the **count** of remaining values


````python
np.random.seed(42)
samples=np.random.normal(50,10,1000)
samples=samples[(samples>=40)&(samples<=60)]
np.count_nonzero(samples)

# OR
filtered_count = np.sum((samples >= 40) & (samples <= 60))
filtered_count
````

##### 7️⃣ Uniform vs Normal comparison

Generate:

* 1000 samples from `uniform(0,1)`
* 1000 samples from `normal(0,1)`
  Compute:
* mean
* standard deviation
  Compare numerically (no plots)


````python
np.random.seed(42)
uniform_dist=np.random.uniform(0,1,1000)
normal_dist=np.random.normal(0,1,1000)
uniform_dist_mean=np.mean(uniform_dist)
uniform_dist_std=np.std(uniform_dist)
normal_dist_mean=np.mean(normal_dist)
normal_dist_std=np.std(normal_dist)

print(f"Uniform Dist Mean :{uniform_dist_mean} and Standard Deviation :{uniform_dist_std}")
print(f"Normal Dist Mean :{normal_dist_mean} and Standard Deviation :{normal_dist_std}")
````

##### 🔟 Monte Carlo π estimation

Use **random sampling** to estimate π:

* Generate 1,000,000 random points
* Points inside unit circle
  📌 No loops
  📌 Pure NumPy


````python
np.random.seed(42)
x= np.random.rand(1000000)
y= np.random.rand(1000000)
r=x**2+y**2
count=np.sum(r<=1.0)
result=4*(count/1000000)
result
````

````python
np.random.seed(42)
points = np.random.rand(1_000_000, 2)
inside = np.sum(np.sum(points**2, axis=1) <= 1)
pi_estimate = 4 * inside / 1_000_000
pi_estimate
````

##### 1️⃣2️⃣ Random walk

Simulate a **1D random walk** of length `1000`:

* Steps = `+1` or `-1`
* Compute:

  * final position
  * max distance from origin


````python
np.random.seed(42)
steps=np.random.choice([-1,1],1000)
walk=np.cumsum(steps)
final_position=walk[-1]
max_distance=np.max(np.abs(walk))
max_distance
````

#####  1️⃣4️⃣ Random weighted matrix

* Generate a `(4,4)` matrix from `uniform(0,1)`
* Multiply each row by random weights from `normal(mean=10, std=2)`
* Normalize final matrix between `0` and `1`


````python
np.random.seed(42)
A=np.random.rand(4,4)
weights=np.random.normal(10,2,4).reshape(4,1)
weighted= A*weights
normalized = (weighted - weighted.min())/(weighted.max()-weighted.min())
normalized
````
